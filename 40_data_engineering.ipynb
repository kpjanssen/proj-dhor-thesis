## Prerequisites

```python
import time
import os
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import TruncatedSVD
from sklearn.feature_extraction.text import TfidfVectorizer
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
file_path = 'data/'
parties = np.load('00_parties.npy')
```

## Topics TF-IDF SVD

```python
dur = time.time()
options = ['imbalanced','oversampling','undersampling','balancedsampling']
#options = ['imbalanced']
ngram         = (1, 2)         # Define bigrams and trigrams (1,1 --> 1.5 were tested 1,2 was best)
minnocc_ratio = 0.001          # Define the required minimal occurance of a word (0.001 -> -.005 were tested 0.001 best)
ncomp_ratio   = 0.03           # Define number of columns (0.11 -> 0.08 were tested 0.03 best)
nwords        = 3              # Define number of words in column header
# -------------------------------------------------------------------------------------~----------------------------------------


# -------------------------------------------------------------------------------------~----------------------------------------
# For each party
for party in parties:
    print(f"{party}") 
    print("-----------------------------------------------------------------------")
    print("-----------------------------------------------------------------------")
    print("-----------------------------------------------------------------------")
    
# -------------------------------------------------------------------------------------~----------------------------------------
# For each option
    for option in options:
        print(f"\n{option}") 
        print("-----------------------------------------------------------------------")
        
# -------------------------------------------------------------------------------------~----------------------------------------
# X_train, train
        X_train_url = file_path + f"10_models/{option}/" + f"{party}_X_train.csv"
        X_train = pd.read_csv(X_train_url)
        
        minnocc = int(minnocc_ratio * len(X_train))

        tfidf = TfidfVectorizer(ngram_range=ngram,min_df=minnocc,use_idf=True)
        tfidf_vect = tfidf.fit_transform(X_train['text'])
        tfidf_columns = tfidf.get_feature_names_out()
        tfidf_df = pd.DataFrame(tfidf_vect.toarray(), columns=tfidf_columns)
        
        ncomp = int(ncomp_ratio * len(tfidf_df.columns))
        
        print(f"rows:{len(X_train)}, minimal occurance:{minnocc}, number of components: {ncomp}")
              
        scaler = StandardScaler()
        scaled = scaler.fit_transform(tfidf_df)

        svd = TruncatedSVD(n_components=ncomp)
        svd_vect = svd.fit_transform(scaled)
        svd_df = pd.DataFrame(svd_vect)
        svd_components = svd.components_
        
        # Plot explained variance using Seaborn
        explained_variance = np.cumsum(svd.explained_variance_ratio_)
        plt.figure(figsize=(14, 6))
        sns.lineplot(x=range(1, ncomp + 1), y=explained_variance, marker='o', color='#093158')
        plt.title('Cumulative Explained Variance')
        plt.xlabel('Number of Components')
        plt.ylabel('Cumulative Explained Variance')
        sns.despine()
        plt.tight_layout()
        plt.show()
        
        for i, component in enumerate(svd_components):
            component_with_names = list(zip(tfidf_df, component))
            sorted_words = sorted(component_with_names, key=lambda x: abs(x[1]), reverse=True)
            top_words = [word for word, _ in sorted_words[:nwords]]
            words = ' '.join(top_words)
            svd_df.rename(columns={i: words}, inplace=True)
        svd_df.columns = svd_df.columns.map(lambda x: ' '.join(sorted(set(x.split()), key=x.split().index)))

        print("X_train")
        X_train = pd.concat([X_train, svd_df], axis=1)
        display(X_train.head(1))  # Print data
        X_train_url = file_path + f"20_models/{option}/" + f"{party}_X_train.csv"
        X_train.to_csv(X_train_url, index=False)    
        
# -------------------------------------------------------------------------------------~----------------------------------------
# X_valid, transform

        X_valid_url = file_path + f"10_models/{option}/" + f"{party}_X_valid.csv"   
        X_valid = pd.read_csv(X_valid_url)    

        tfidf_vect = tfidf.transform(X_valid['text'])
        tfidf_df = pd.DataFrame(tfidf_vect.toarray(), columns=tfidf_columns)

        scaled = scaler.transform(tfidf_df)

        svd_vect = svd.transform(scaled)
        svd_df = pd.DataFrame(svd_vect)
        svd_components = svd.components_

        for i, component in enumerate(svd_components):
            component_with_names = list(zip(tfidf_df, component))
            sorted_words = sorted(component_with_names, key=lambda x: abs(x[1]), reverse=True)
            top_words = [word for word, _ in sorted_words[:nwords]]
            words = ' '.join(top_words)
            svd_df.rename(columns={i: words}, inplace=True)
        svd_df.columns = svd_df.columns.map(lambda x: ' '.join(sorted(set(x.split()), key=x.split().index)))

        print("X_valid")
        X_valid = pd.concat([X_valid, svd_df], axis=1)
        display(X_valid.head(1))  # Print data
        X_valid_url = file_path + f"20_models/{option}/" + f"{party}_X_valid.csv"   
        X_valid.to_csv(X_valid_url, index=False)         
        

# -------------------------------------------------------------------------------------~----------------------------------------
# Final model sets
        if option == 'imbalanced':
            print("Final model")
# -------------------------------------------------------------------------------------~----------------------------------------
# X_trainvalid
            X_trainvalid_url = file_path + f"30_models_imbalanced/" + f"{party}_X_trainvalid.csv"
            X_trainvalid = pd.read_csv(X_trainvalid_url)

            minnocc = int(minnocc_ratio * len(X_trainvalid))

            tfidf = TfidfVectorizer(ngram_range=ngram,min_df=minnocc,use_idf=True)
            tfidf_vect = tfidf.fit_transform(X_trainvalid['text'])
            tfidf_columns = tfidf.get_feature_names_out()
            tfidf_df = pd.DataFrame(tfidf_vect.toarray(), columns=tfidf_columns)

            ncomp = int(ncomp_ratio * len(tfidf_df.columns))

            print(f"rows:{len(X_trainvalid)}, minimal occurance:{minnocc}, number of components: {ncomp}")

            scaler = StandardScaler()
            scaled = scaler.fit_transform(tfidf_df)

            svd = TruncatedSVD(n_components=ncomp)
            svd_vect = svd.fit_transform(scaled)
            svd_df = pd.DataFrame(svd_vect)
            svd_components = svd.components_

            # Plot explained variance using Seaborn
            explained_variance = np.cumsum(svd.explained_variance_ratio_)
            plt.figure(figsize=(14, 6))
            sns.lineplot(x=range(1, ncomp + 1), y=explained_variance, marker='o', color='#093158')
            plt.title('Cumulative Explained Variance')
            plt.xlabel('Number of Components')
            plt.ylabel('Cumulative Explained Variance')
            sns.despine()
            plt.tight_layout()
            plt.show()

            for i, component in enumerate(svd_components):
                component_with_names = list(zip(tfidf_df, component))
                sorted_words = sorted(component_with_names, key=lambda x: abs(x[1]), reverse=True)
                top_words = [word for word, _ in sorted_words[:nwords]]
                words = ' '.join(top_words)
                svd_df.rename(columns={i: words}, inplace=True)
            svd_df.columns = svd_df.columns.map(lambda x: ' '.join(sorted(set(x.split()), key=x.split().index)))

            print("X_trainvalid")
            X_trainvalid = pd.concat([X_trainvalid, svd_df], axis=1)
            display(X_trainvalid.head(1))  # Print data
            X_trainvalid_url = file_path + f"40_models_imbalanced/" + f"{party}_X_trainvalid.csv"
            X_trainvalid.to_csv(X_trainvalid_url, index=False)    

        
# -------------------------------------------------------------------------------------~----------------------------------------
# X_test

            X_test_url = file_path + f"30_models_imbalanced/" + f"{party}_X_test.csv"   
            X_test = pd.read_csv(X_test_url)    

            tfidf_vect = tfidf.transform(X_test['text'])
            tfidf_df = pd.DataFrame(tfidf_vect.toarray(), columns=tfidf_columns)

            scaled = scaler.transform(tfidf_df)

            svd_vect = svd.transform(scaled)
            svd_df = pd.DataFrame(svd_vect)
            svd_components = svd.components_

            for i, component in enumerate(svd_components):
                component_with_names = list(zip(tfidf_df, component))
                sorted_words = sorted(component_with_names, key=lambda x: abs(x[1]), reverse=True)
                top_words = [word for word, _ in sorted_words[:nwords]]
                words = ' '.join(top_words)
                svd_df.rename(columns={i: words}, inplace=True)
            svd_df.columns = svd_df.columns.map(lambda x: ' '.join(sorted(set(x.split()), key=x.split().index)))

            print("X_test")
            X_test = pd.concat([X_test, svd_df], axis=1)
            display(X_test.head(1))  # Print data
            X_test_url = file_path + f"40_models_imbalanced/" + f"{party}_X_test.csv"   
            X_test.to_csv(X_test_url, index=False) 
        
        print("\n\n\n\n\n")
        
# -------------------------------------------------------------------------------------~----------------------------------------
print('\n---------------------------------------------------------------------------------------------------------------------')
print(f"Code duration: {round((time.time()  - dur),3)} seconds")      
```
